```python
from langchain_groq import ChatGroq
from langchain_huggingface import HuggingFaceEndpointEmbeddings
from dotenv import load_dotenv

load_dotenv()

model = ChatGroq(model='openai/gpt-oss-120b')

em_model = HuggingFaceEndpointEmbeddings(model='Qwen/Qwen3-Embedding-4B')
print('a')
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_classic.document_loaders import PyPDFLoader

file = "Hands-on-Machine-Learning.pdf"
print('b')
loader = PyPDFLoader(file)

data = loader.load()

from langchain_chroma import Chroma 
path = './chroma_store'
page_store = Chroma(
    embedding_function=em_model,
    collection_name='page_store',
    persist_directory=path
)
chunk_store = Chroma(
    embedding_function=em_model,
    collection_name='chunk_store',
    persist_directory=path
)
```

```python
page_store.add_documents(data)
```

```python
page_store.clear()
```

```python
from langchain_pymupdf4llm import PyMuPDF4LLMLoader
file='Hands-on-Machine-Learning.pdf'
loader = PyMuPDF4LLMLoader(
    file_path=file,
    mode="page",
)
```

```python
docs = []
i=0
for x in loader.lazy_load():
    docs.append(x)
    if i%50==0:
        print(f'page {i} done')
    i+=1
```
