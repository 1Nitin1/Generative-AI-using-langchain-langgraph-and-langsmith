```python

# Simple project to get similar lines to a user query from document.
from langchain_huggingface import HuggingFaceEmbeddings

model = HuggingFaceEmbeddings(model_name="sentence-transformers/paraphrase-TinyBERT-L6-v2")

document = [
    "The quick brown fox jumps over the lazy dog.",
    "Artificial intelligence is transforming industries worldwide.",
    "I enjoy hiking in the mountains during winter.",
    "Music has the power to change our mood instantly.",
    "The library was silent except for the sound of turning pages.",
    "She brewed a fresh cup of coffee before starting work.",
    "Space exploration inspires curiosity about the unknown.",
    "The cat stretched lazily under the warm sunlight.",
    "Learning new programming languages expands problem-solving skills.",
    "Rain tapped gently against the window as evening fell."
]

embed_doc=[model.embed_query(x) for x in document]



from sklearn.metrics.pairwise import cosine_similarity

user_query=input("Ask your question: ")
q_embed=model.embed_query(user_query)

sim = cosine_similarity([q_embed],embed_doc)[0]

arr = sorted(list(enumerate(sim)),key=(lambda x:x[1]))[-1]

print(f"The match score is {arr[1]}")

print(document[arr[0]])
```

```python
## Structured output

# it means getting data is structured format like json.
# llms that can - with_structured_output and can't - output parsers
# typeddict, pydantic, json_schema
# this code will work with openai, gemini, etc

# TypedDict, Annotated, Optional usages.
# Annotated gives details to llm. 
# optional is used when we might not get that output if not available
# we can't have default values in typeddict, also no data validation
from typing import TypedDict, Annotated, Optional
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
from dotenv import load_dotenv

#schema
class Review(TypedDict):
    
    key_themes: Annotated[list[str],'write all the key themes discussed in the review in a list']
    summary:Annotated[str,"a brief summary of the review"]
    sentiment:Annotated[str,"return sentiment of the review, either negative, positive or neutral"]
    pros:Annotated[Optional[list[str]],'write all pros in a list']

load_dotenv()

llm = HuggingFaceEndpoint(
    repo_id="deepseek-ai/DeepSeek-V4.1-Flash",
)
model = ChatHuggingFace(llm=llm)

str_model = model.with_structured_output(Review)

result = str_model.invoke("""the hardware is great, but the software feels bloated. there are too many 
pre-installed apps that i can't remove. also, the ui looks outdated compared to other brands. hoping 
for a software update to fix this.""")
```

```python
# Pydantic
# We get inbuilt data validation
# also can set default values
# does implicit data conversion if possible
# can use optional
# Field : can set default values, ranges, descriptions, regex, etc
# can convert to dict using dict(result.content)
# can convert to json using model_dump_json()
from pydantic import BaseModel, EmailStr, Field 
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
from typing import Optional
from dotenv import load_dotenv

#schema
class Student(BaseModel):
    name: str = 'nitin'
    age: Optional[int] = None
    email: EmailStr
    cgpa: float = Field(gt=0,lt=10,default=5,description='decimal value representing cgpa of the student')

load_dotenv()

llm = HuggingFaceEndpoint(
    repo_id="deepseek-ai/DeepSeek-R1",
)
model = ChatHuggingFace(llm=llm)

str_model = model.with_structured_output(Student)

result = str_model.invoke("""give me a random name and age of a person""")
```

```python
# structured output parser
# it is provided a predefined field schema
# we use responseSchema
# we can't do data validation

from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import JsonOutputParser
from pydantic import BaseModel, Field
from typing import Annotated

from dotenv import load_dotenv

load_dotenv()

class Facts(BaseModel):
    fact1: Annotated[str , Field(description="This is the first fact")]
    fact2: Annotated[str , Field(description="This is the second fact")]
    fact3: Annotated[str , Field(description="This is the third fact")]

model = ChatGroq(model='openai/gpt-oss-120b')

parser = JsonOutputParser(pydantic_object=Facts)

template = PromptTemplate(
    template = " write 3 facts about {topic},\n {format_instruction}",
    input_variables=[],
    partial_variables={'format_instruction':parser.get_format_instructions()}
)

chain = template | model | parser

chain.invoke({'topic':'black hole'})
```

```python
#Simple + Sequence chains
from langchain_groq import ChatGroq
from dotenv import load_dotenv
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
from langchain_core.runnables import RunnableSequence

load_dotenv()

model = ChatGroq(model='openai/gpt-oss-120b')

template1 = PromptTemplate(
    template="Give a detailed report on {topic}",
    input_variables = ['topic']
)

template2 = PromptTemplate(
    template="Give a 5 point summary on the below report \n {text}",
    input_variables = ['text']
)

parser = StrOutputParser()

chain = RunnableSequence(template1, model, parser, template2, model, parser)

result = chain.invoke({'topic':"block hole"})

result
```

```python
# parallel chains
from langchain_groq import ChatGroq
from dotenv import load_dotenv
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
from langchain_core.runnables import RunnableSequence, RunnableParallel

load_dotenv()

model = ChatGroq(model='openai/gpt-oss-120b')

template1 = PromptTemplate(
    template="Give a 5 line on {topic}",
    input_variables = ['topic']
)

template2 = PromptTemplate(
    template="Give 3 interesting facts on {topic}",
    input_variables = ['topic']
)

template3 = PromptTemplate(
    template="merge \n summary -> {summary} \n and \n facts -> {facts} \n and return to me",
    input_variables = ['summary','facts']
)

parser = StrOutputParser()

parallel_chain = RunnableParallel({
    'summary':RunnableSequence(template1,model,parser),
    'facts':RunnableSequence(template2,model,parser)
})

chain = RunnableSequence(parallel_chain,template3,model,parser)

result = chain.invoke({'topic':"block hole"})

result
```

```python
# conditional chains
from langchain_groq import ChatGroq
from dotenv import load_dotenv
from langchain_core.output_parsers import StrOutputParser, JsonOutputParser
from pydantic import BaseModel, Field
from typing import Literal, Annotated
from langchain_core.prompts import PromptTemplate
from langchain_core.runnables import RunnableSequence, RunnableBranch, RunnableLambda

load_dotenv()

model = ChatGroq(model='openai/gpt-oss-120b')

class Class(BaseModel):
    classification: Annotated[Literal['pos','neg'],Field(description="classify as either pos or neg")]
    review: Annotated[str,Field(description="the same input review")]

parser1 = StrOutputParser()
parser2 = JsonOutputParser(pydantic_object=Class) 


template1 = PromptTemplate(
    template="Classify the below review as positive or negative: \n {review} \n {format_instructions}",
    input_variables = ['review'],
    partial_variables={'format_instructions':parser2.get_format_instructions()}
)

template2 = PromptTemplate(
    template="Answer the below positive review with a thanks message. \n {review}",
    input_variables = ['review']
)

template3 = PromptTemplate(
    template="Answer the below negative review with a sorry message. \n {review}",
    input_variables = ['review']
)



con_chain = RunnableBranch(
    (lambda x:x['classification']=='pos',RunnableSequence(template2,model,parser1)),
    (lambda x:x['classification']=='neg',RunnableSequence(template3,model,parser1)),
    RunnableLambda(lambda x: "No such classification available")
)

chain = RunnableSequence(template1, model, parser2, con_chain)

result = chain.invoke({'review':"This is the worst fast food shop in whole India"})

result
```

```python
# 3. document structure based splitter : like code and markdown
from langchain_text_splitters import RecursiveCharacterTextSplitter, Language

text ="""class Student:
    '''
    A simple Student class to demonstrate:
    1. Storing basic student info
    2. Functions for updating and displaying details
    '''

    def __init__(self, name, age):
        # Constructor initializes student with name and age
        self.name = name
        self.age = age

    def update_age(self, new_age):
        # Function to update the student's age
        self.age = new_age
        print(f"Age updated to {self.age}")

    def display_info(self):
        # Function to display student details
        print(f"Student Name: {self.name}")
        print(f"Student Age: {self.age}")


# Example usage
student1 = Student("Nitin", 21)   # Create a student object
student1.display_info()           # Show initial details
student1.update_age(22)           # Update age
student1.display_info()           # Show updated details
"""
splitter = RecursiveCharacterTextSplitter.from_language(
    language=Language.PYTHON,
    chunk_size=200,
    chunk_overlap=10,
)

res = splitter.split_text(text)
```

```python
from langchain_groq import ChatGroq
from langchain_huggingface import HuggingFaceEndpointEmbeddings
from dotenv import load_dotenv
from youtube_transcript_api import YouTubeTranscriptApi
from langchain_chroma import Chroma
import os
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import JsonOutputParser, StrOutputParser
from pydantic import BaseModel, Field
from typing import List,Annotated
from langchain_core.runnables import RunnableSequence, RunnableLambda, RunnableParallel, RunnablePassthrough

#Load .env 
load_dotenv()

# LLM answer template
class Answer(BaseModel):
    result:Annotated[str,Field(description='This is the main output result')]


# LLM answer json parser
parser2 = JsonOutputParser(pydantic_object=Answer)

#LLM answer string parser
parser = StrOutputParser()

# LLM prompt template
temp = PromptTemplate(
    template = "Answer the query in English only -> {query} <- using the content give below, if there is no context about the query in the data provided, do not hallucinate and return that it's not in the data. {format_instructions}: \n {text_data}",
    input_variables=['query','text_data'],
    partial_variables={'format_instructions':parser2.get_format_instructions()}
)
    
print('a')
# LLM chat model
model = ChatGroq(model='openai/gpt-oss-120b')

# Embedding model
embedding_model = HuggingFaceEndpointEmbeddings(
    model="Qwen/Qwen3-Embedding-8B"
)

print('b')
# URL to video_id extraction
url = input('Enter the url :')
video_id = (url.split('?')[0]).split('/')[-1]
if video_id=="watch":
    video_id="".join(x for x in url.split('=')[1:])

# Path of store
folder_path = os.path.join(os.getcwd(),video_id)

# Un-initialized store
store = None
print('c')
# Storage task pipeline works if there is no store for the input video
if not os.path.isdir(folder_path):
    # Load timestamp data
    api = YouTubeTranscriptApi()
    data=api.fetch(video_id,languages=['en','hi'])

    # merge in one
    transcript = " ".join(chunk.text for chunk in data)

    # use splitter to create documents
    from langchain_text_splitters import RecursiveCharacterTextSplitter
    splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
    chunks = splitter.create_documents([transcript])

    # Create the store
    store = Chroma(
        embedding_function=embedding_model,
        collection_name='sample',
        persist_directory=video_id
    )

    # add all the chunks(documents)
    ids = store.add_documents(chunks)
else:
    store = Chroma(
        embedding_function=embedding_model,
        collection_name='sample',
        persist_directory=video_id
    )

#create retriever
retriever = store.as_retriever(search_type="similarity",search_kwargs={'k':3})

user_input = ""
while user_input != 'exit':
    user_input=input('Query :')
    p_chain = RunnableParallel({
        'query':RunnablePassthrough(),
        'text_data':RunnableSequence(retriever, RunnableLambda(lambda data: " ".join(x.page_content for x in data)))
    })

    chain = RunnableSequence(p_chain, temp , model , RunnableLambda(lambda x:x.content) , parser , parser2 , RunnableLambda(lambda x: x['result']))
    result = chain.invoke(user_input)
    
    print(f'AI : {result}')
```

```python
# custom tool making
# 3 ways - use @tool, use StructuredTool(pydantic support), use Basetool
from langchain_core.tools import tool

@tool
def multiply(a:int,b:int)->int:
    '''Multiply two numbers'''
    return a*b
```

```python
multiply.invoke({'a':3,'b':100})
```

```python
# using StructuredTool
from langchain_core.tools import StructuredTool
from pydantic import BaseModel, Field

class MultiplyInput(BaseModel):
    a:int=Field(description='This is the first integer')
    b:int=Field(description='This is the second integer')

def multiply1(a:int,b:int)->int:
    return a*b

my_tool = StructuredTool.from_function(
    func=multiply1,
    name='multiply_tool',
    description='multiply two numbers',
    args_schema=MultiplyInput
)
```

```python
my_tool.invoke({'a':7,'b':-7})
```

```python
my_tool.name, my_tool.description, multiply.args_schema.model_json_schema()
```

```python
#using BaseTool
from langchain_core.tools import BaseTool
from typing import Type
class MultiplyTool(BaseTool):
    name:str='multiply_tool'
    description:str='multiply two numbers'
    args_schema:Type[BaseModel]=MultiplyInput

    def _run(self,a:int,b:int)->int:
        return a*b
```

```python
MultiplyTool().invoke({'a':123,'b':-10})
```

```python
from langchain_core.tools import tool
from langchain_groq import ChatGroq
from dotenv import load_dotenv

load_dotenv()

model = ChatGroq(model='openai/gpt-oss-120b')

@tool
def multiply_tool(a:int,b:int)->int:
    '''tool to perform Multiplication of two numbers'''
    return a*b

model=model.bind_tools([multiply_tool])
res = model.invoke('check the tools first. If there is not relevant tool than say so. \n multiplication of 57933 and 5')
```

```python
# create_react_agent and AgentExecutor
from langchain_core.tools import tool
from langchain.agents import create_agent
from langchain_core.messages import HumanMessage, AIMessage
import yagmail
import json
from langchain_groq import ChatGroq
from dotenv import load_dotenv

load_dotenv()

llm = ChatGroq(model='openai/gpt-oss-120b')

@tool
def send_email(to:str,subject:str,contents:str)->str:
    '''Tool for sending emails'''
    yag=yagmail.SMTP('nitinkumarbaranawal@gmail.com','wvsx ezal uoph upeh')
    yag.send(
        to,
        subject,
        contents
    )
    return 'Message sent successfully'

@tool
def find_email(name:str)->str:
    '''Tool for finding email from database using name of person'''
    with open('contacts.json','r') as f:
        contacts = json.load(f)
        if not contacts.get(name.title()):
            email = input(f'Email for {name} is not in database. Please enter manually : ')
            contacts[name.title()]=email;
            with open('contacts.json','w') as f:
                json.dump(contacts,f,indent=4)
        else:
            return contacts.get(name.title())
        return email

agent = create_agent(
    model=llm,
    tools=[send_email,find_email]
)
```

```python
from langchain_core.messages import AIMessage, ToolMessage, HumanMessage
from langchain_core.prompts import PromptTemplate
prompt = PromptTemplate(
    template = 'I am Anonymous. reAct to my order below \n {data}',
    input_variables=['data']
)

data = input('User :')
print()


for chunk in agent.stream(
    {"messages":[HumanMessage(prompt.invoke({'data':data}).text)]}
):
    
    if "model" in chunk:
        for msg in chunk["model"]["messages"]:
            if isinstance(msg, AIMessage):
                
                if "content" in msg.additional_kwargs:
                    print("AI reasoning:", msg.additional_kwargs["content"])
                
                if "tool_calls" in msg.additional_kwargs:
                    for tc in msg.additional_kwargs["tool_calls"]:
                        print(f"AI wants to call tool: {tc['function']['name']} with args {tc['function']['arguments']}")
                if msg.content:
                    print("AI:", msg.content)

    elif "tools" in chunk:
        for msg in chunk["tools"]["messages"]:
            print(f"Tool {msg.name} returned: {msg.content}")
    print()

```

```python
from langchain_groq import ChatGroq
from kokoro import KPipeline
import whisper
import speech_recognition as sr
from langchain_core.runnables import RunnableLambda, RunnableParallel, RunnableBranch
from typing import List, Dict, Optional, Annotated
from pydantic import BaseModel, Field
from langchain_core.messages import AIMessage, HumanMessage, SystemMessage
from langchain_core.output_parsers import StrOutputParser, JsonOutputParser, PydanticOutputParser
from langchain_core.prompts import PromptTemplate
import sounddevice as sd
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain_core.tools import tool
import json
import yagmail


load_dotenv()

model1 = ChatGroq(model='openai/gpt-oss-120b')

pipe = KPipeline(lang_code='a')
def kokoro_tts(msg:str,voice='af_bella'):
    generator = pipe(msg,voice)
    for _,_,audio in generator:
        sd.play(audio,samplerate=24000)
        sd.wait()

model2 = RunnableLambda(lambda msg:kokoro_tts(msg))

modelk = whisper.load_model('medium')

model3 = RunnableLambda(lambda _: modelk.transcribe('output.wav')['text'])



tool
def send_email(to:str,subject:str,contents:str)->str:
    '''Tool for sending emails in behalf of Nitin Baranwal'''
    yag=yagmail.SMTP('nitinkumarbaranawal@gmail.com','wvsx ezal uoph upeh')
    yag.send(
        to,
        subject,
        contents
    )
    return 'Message sent successfully'

@tool
def find_email(name:str)->str:
    '''Tool for finding email from database using name of person'''
    with open('contacts.json','r') as f:
        contacts = json.load(f)
        if not contacts.get(name.title()):
            email = input(f'Email for {name} is not in database. Please enter manually : ')
            contacts[name.title()]=email;
            with open('contacts.json','w') as f:
                json.dump(contacts,f,indent=4)
        else:
            return contacts.get(name.title())
        return email

agent = create_agent(
    model=model1,
    tools=[send_email,find_email]
)

r = sr.Recognizer()
r.energy_threshold = 300
r.pause_threshold = 4
def listen_speech():
    with sr.Microphone() as source:
        audio = r.listen(source)

    with open('output.wav','wb') as f:
        f.write(audio.get_wav_data())
```

```python
class LLMOutput(BaseModel):
    description: Annotated[str,Field(description='It is the description of output')]
    speech: Annotated[str,Field(description='It is the actual sentences that you will speak to the master')]

json_parser = PydanticOutputParser(pydantic_object=LLMOutput)

template1 = PromptTemplate(
    template="""You are a polite assistant that summarizes the agent's work.

- If the agent.invoke output shows tool calls (e.g., find_email, send_email, search, etc.), respond like a courteous servant. 
  - Confirm the task is completed.
  - List the steps taken in order, in a polite and concise way.
  - End with a reassuring line such as "The work has been completed successfully."

- If the agent.invoke output shows no tool calls and is just a normal query/answer, respond like a knowledgeable teacher. 
  - Explain the answer clearly.
  - Use an instructive, helpful tone.
  - Provide context or examples if useful.

Always keep the response polite, clear, and role‑appropriate.
\n output -> {data} \n {format_instructions}""",
    input_variables=['data'],
    partial_variables={'format_instructions':json_parser.get_format_instructions()}
)

groq_parser = RunnableLambda(lambda x: x.content)
formatter1 = RunnableLambda(lambda x:{'messages':HumanMessage("I am your master Nitin. "+x)})
formatter2 = RunnableLambda(lambda x:{'data':x})
```

```python
i=2
import time
while True:
    print('Listening...')
    start=time.time()
    listen_speech()
    
    chain = model3 | formatter1 | agent | formatter2 | template1 | model1 | groq_parser | json_parser | RunnableLambda(lambda x:x.speech) | model2
    
    a = chain.invoke('output.wav')
    if i==2:
        break
    i+=1
    
print(time.time()-start)
```
